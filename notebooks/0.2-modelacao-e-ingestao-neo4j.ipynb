### Execução de Consultas Declarativas Cypher
**Consulta 1:** Responder à Questão 7.1 da lista de exercícios: disciplinas que possuem Algoritmos como pré-requisito.

```python
query_algoritmos = '''
MATCH (alg:Componente {codigo: 'TADS01'})-[:E_PRE_REQUISITO_DE]->(dep:Componente)
RETURN dep.codigo AS Codigo, dep.nome AS Disciplina, dep.semestre AS Semestre
ORDER BY dep.semestre
'''
print('Cypher Query Preparada:\n', query_algoritmos)
```

**Consulta 2:** Travessia profunda de múltiplos saltos (*k-hop traversal* de 1 a 3 níveis).

```python
query_multi_hop = '''
MATCH path = (alg:Componente {codigo: 'TADS01'})-[:E_PRE_REQUISITO_DE*1..3]->(alvo:Componente)
RETURN alvo.codigo AS Codigo, alvo.nome AS Disciplina, length(path) AS Saltos
ORDER BY Saltos, alvo.semestre
'''
print('Cypher Multi-Hop Query:\n', query_multi_hop)
```
