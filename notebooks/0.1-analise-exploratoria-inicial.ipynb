```python
import sys
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Carregar módulos do projeto
sys.path.append(str(Path('.').resolve().parent))
from src.config import PROCESSED_DATA_DIR

# Leitura dos dados higienizados
df_disc = pd.read_csv(PROCESSED_DATA_DIR / 'nodes_componentes.csv')
df_pre = pd.read_csv(PROCESSED_DATA_DIR / 'edges_prerequisitos.csv')
df_disc.head(10)
```

### 1. Estatísticas Descritivas e Distribuição de Carga Horária (`describe()`)
Avaliação de médias, desvios e amplitudes dos dados curriculares conforme `ds06.pdf`.

```python
print('--- Sumário Estatístico da Carga Horária e Semestres ---')
df_disc[['ch', 'semestre']].describe()
```

### 2. Distribuição por Período Letivo e Áreas de Conhecimento

```python
plt.figure(figsize=(10, 4))
sns.countplot(data=df_disc, x='semestre', palette='viridis', edgecolor='black')
plt.title('Distribuição de Disciplinas por Período Curricular', fontsize=12, weight='bold')
plt.xlabel('Período Letivo (1º ao 6º Semestre)')
plt.ylabel('Quantidade de Componentes')
plt.show()
```

```python
plt.figure(figsize=(10, 4))
order = df_disc['area'].value_counts().index
sns.countplot(data=df_disc, y='area', order=order, palette='crest', edgecolor='black')
plt.title('Concentração Curricular por Área de Formação', fontsize=12, weight='bold')
plt.xlabel('Número de Disciplinas')
plt.ylabel('Área')
plt.show()
```
